# Grade 12 · Unit 7 — Communicating with data

> **SYNTHETIC** district table.

```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

def data_file(dataset, filename):
    """Find datasets/<dataset>/clean/<filename> by walking up from this folder."""
    here = Path.cwd().resolve()
    for folder in [here, *here.parents]:
        p = folder / "datasets" / dataset / "clean" / filename
        if p.exists():
            return p
    raise FileNotFoundError(f"Could not find {filename}. Open this notebook from inside the data-science-nepal folder.")

districts = pd.read_csv(data_file("census-districts", "districts_synthetic.csv"))
districts["abroad_per_1000"] = 1000 * districts["absentee_population"] / districts["population"]
by_belt = districts.groupby("ecological_belt")["abroad_per_1000"].median().sort_values()
assert len(by_belt) == 3
by_belt.round(1)
```

## Topic title vs action title
* Topic: 'Abroad per 1,000 by belt'
* Action: the *message*, e.g. 'In this invented table, hill districts have the most people abroad per 1,000'. Compute first, then write the message to match the data.

```python
top = by_belt.index[-1]
fig, ax = plt.subplots(figsize=(6, 3))
by_belt.plot(kind="barh", ax=ax, color=["#bbbbbb"] * 2 + ["#c0392b"])
ax.set_title(f"{top} districts have the most people abroad per 1,000 (SYNTHETIC)", loc="left", fontsize=10)
ax.set_xlabel("Median per 1,000 people"); ax.set_ylabel("")
fig.text(0.01, -0.05, "Source: synthetic practice table, Data Science Nepal (CC BY 4.0). Limit: invented numbers.", fontsize=7)
import tempfile, os
path = os.path.join(tempfile.mkdtemp(), "story.png")
fig.savefig(path, bbox_inches="tight"); assert os.path.getsize(path) > 1000
plt.show()
```
