# Grade 9 · Unit 6 — Charts that work

> **SYNTHETIC** district data.

```python
from pathlib import Path
import pandas as pd

def data_file(dataset, filename):
    """Find datasets/<dataset>/clean/<filename> by walking up from this folder."""
    here = Path.cwd().resolve()
    for folder in [here, *here.parents]:
        p = folder / "datasets" / dataset / "clean" / filename
        if p.exists():
            return p
    raise FileNotFoundError(f"Could not find {filename}. Open this notebook from inside the data-science-nepal folder.")

districts = pd.read_csv(data_file("census-districts", "districts_synthetic.csv"))
schools = pd.read_csv(data_file("census-districts", "schools_synthetic.csv"))
print(districts.shape, schools.shape)
import matplotlib.pyplot as plt
```

## Histogram — the shape of one numerical variable
**Predict:** will literacy be spread evenly or bunch up?

```python
plt.hist(districts["literacy_pct"], bins=10, edgecolor="black")
plt.xlabel("Literacy (%)"); plt.ylabel("Number of districts")
plt.title("Literacy in 77 districts (SYNTHETIC)")
plt.show()
```

## Bar chart — categories

```python
counts = districts["ecological_belt"].value_counts()
plt.bar(counts.index, counts.values)
plt.ylabel("Number of districts"); plt.title("Districts by ecological belt (SYNTHETIC)")
plt.show()
```

## Box plot — compare groups

```python
groups = [g["literacy_pct"].values for _, g in districts.groupby("ecological_belt")]
labels = sorted(districts["ecological_belt"].unique())
plt.boxplot(groups, tick_labels=labels)
plt.ylabel("Literacy (%)"); plt.title("Literacy by belt (SYNTHETIC)")
plt.show()
```

## Scatter plot — two numerical variables
**Predict:** do more populous districts have higher literacy?

```python
plt.scatter(districts["population"] / 1000, districts["literacy_pct"])
plt.xlabel("Population (thousands)"); plt.ylabel("Literacy (%)")
plt.title("Population vs literacy (SYNTHETIC)")
plt.show()
```
