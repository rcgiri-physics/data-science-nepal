# Grade 9 · Unit 7 — Percent, rate and per-capita

> **SYNTHETIC** district and school tables.

```python
from pathlib import Path
import pandas as pd

def data_file(dataset, filename):
    """Find datasets/<dataset>/clean/<filename> by walking up from this folder."""
    here = Path.cwd().resolve()
    for folder in [here, *here.parents]:
        p = folder / "datasets" / dataset / "clean" / filename
        if p.exists():
            return p
    raise FileNotFoundError(f"Could not find {filename}. Open this notebook from inside the data-science-nepal folder.")

districts = pd.read_csv(data_file("census-districts", "districts_synthetic.csv"))
schools = pd.read_csv(data_file("census-districts", "schools_synthetic.csv"))
print(districts.shape, schools.shape)
```

## Counts vs rates
**Predict:** is the district with the most students abroad also the one with the highest *rate*?

```python
d = districts.copy()
d["abroad_rate_per_1000"] = 1000 * d["absentee_population"] / d["population"]
top_count = d.sort_values("absentee_population", ascending=False).iloc[0]["district"]
top_rate = d.sort_values("abroad_rate_per_1000", ascending=False).iloc[0]["district"]
print("Highest count:", top_count, "| Highest rate:", top_rate)
```

## Shares and the denominator

```python
m = d.merge(schools, on="district_id")
m["girls_share_pct"] = 100 * m["enrolment_girls"] / (m["enrolment_girls"] + m["enrolment_boys"])
m["enrolled_per_100_people"] = 100 * (m["enrolment_girls"] + m["enrolment_boys"]) / m["population"]
m[["district", "girls_share_pct", "enrolled_per_100_people"]].round(1).head()
```

## Percent change vs percentage points

```python
before, after = 60.0, 66.0
points = after - before
relative = 100 * (after - before) / before
assert points == 6.0 and round(relative, 1) == 10.0
print(f"{points} percentage points = {relative:.1f}% relative change")
```
