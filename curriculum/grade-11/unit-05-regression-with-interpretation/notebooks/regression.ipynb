> **SYNTHETIC** health table (relationships were built into the generator).

```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

def data_file(dataset, filename):
    """Find datasets/<dataset>/clean/<filename> by walking up from this folder."""
    here = Path.cwd().resolve()
    for folder in [here, *here.parents]:
        p = folder / "datasets" / dataset / "clean" / filename
        if p.exists():
            return p
    raise FileNotFoundError(f"Could not find {filename}. Open this notebook from inside the data-science-nepal folder.")

h = pd.read_csv(data_file("health-synthetic", "health_synthetic.csv"))
print(h.shape)
```

## Fit birth weight from wealth and visits

```python
X = np.column_stack([np.ones(len(h)), h["wealth_quintile"], h["antenatal_visits"]])
y = h["birth_weight_kg"].to_numpy()
coef, *_ = np.linalg.lstsq(X, y, rcond=None)
pred = X @ coef
resid = y - pred
r2 = 1 - (resid ** 2).sum() / ((y - y.mean()) ** 2).sum()
assert coef.shape == (3,) and 0 <= r2 <= 1
print(dict(intercept=round(coef[0], 3), per_wealth_step=round(coef[1], 3), per_visit=round(coef[2], 3)), "R² =", round(r2, 3))
```

**Interpretation (practise saying it):** 'Among mothers with the same number of visits, each step up in wealth quintile goes with about … kg higher birth weight.'

## Residuals

```python
plt.scatter(pred, resid, s=6, alpha=0.4); plt.axhline(0, color="red")
plt.xlabel("predicted (kg)"); plt.ylabel("residual (kg)"); plt.show()
```
